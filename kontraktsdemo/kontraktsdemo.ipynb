```python
import json
import random
import time
import html
from copy import deepcopy
from getpass import getpass

import pandas as pd
from IPython.display import display, HTML
from openai import OpenAI

import demo_core as core

# Bare nye API-kall brukes. Tidligere modellsvar er ikke nødvendig.
core.MODE = "api"
core.CALLS = 0

api_key = getpass("Lim inn OpenAI API-nøkkelen din: ").strip()

try:
    if not api_key:
        raise ValueError("API-nøkkelen er tom.")

    core.client = OpenAI(
        api_key=api_key,
        timeout=120,
        max_retries=1,
    )
finally:
    del api_key

# Disse navnene brukes også av cellene som lager oppsummeringen.
client = core.client
MODEL = core.MODEL
REASONING_EFFORT = core.REASONING_EFFORT
CATS = list(core.CATS)
CATEGORIES = core.CATEGORIES

demo_pool = json.loads(
    (core.BASE / "data/demo_reserve.json").read_text(
        encoding="utf-8"
    )
)

if not demo_pool:
    raise ValueError("Datafilen inneholder ingen reservekontrakter.")

print(f"Modell: {MODEL}")
print(f"Kategorier: {len(CATS)}")
print(f"Tilgjengelige reservekontrakter: {len(demo_pool)}")
```
